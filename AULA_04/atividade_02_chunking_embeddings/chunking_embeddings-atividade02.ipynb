```python
import pymupdf4llm

def extrair_pdf_para_markdown(caminho_pdf: str) -> str:
    """Converte um PDF em texto Markdown preservando estrutura."""
    md_text = pymupdf4llm.to_markdown(caminho_pdf)
    return md_text
```

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    NLTKTextSplitter,
    MarkdownHeaderTextSplitter,
)
```

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

def contar_tokens(texto: str) -> int:
    return len(encoding.encode(texto))
```

```python
def gerar_chunks_teste_1_a_4(texto: str, chunk_size: int):
    splitter = CharacterTextSplitter(
        separator="",            # corta puramente por quantidade de caracteres
        chunk_size=chunk_size,
        chunk_overlap=0,
        length_function=len,
    )
    return splitter.split_text(texto)
```

```python
def gerar_chunks_teste_5_e_6(texto: str, chunk_overlap: int):
    splitter = CharacterTextSplitter(
        separator="",
        chunk_size=500,
        chunk_overlap=chunk_overlap,   # 50 no teste 5, 200 no teste 6
        length_function=len,
    )
    return splitter.split_text(texto)
```

```python
def gerar_chunks_teste_7(texto: str):
    splitter = CharacterTextSplitter(
        separator="\n\n",       # quebra de parágrafo
        chunk_size=100000,      # tamanho bem alto: não queremos que ele corte por tamanho,
        chunk_overlap=0,        # só queremos que ele use o \n\n como unidade
    )
    return splitter.split_text(texto)
```

```python
def gerar_chunks_teste_8(texto: str):
    splitter_sentencas = NLTKTextSplitter()
    sentencas = splitter_sentencas.split_text(texto)
    import nltk
    sentencas = nltk.sent_tokenize(texto)

    chunks = []
    for i in range(0, len(sentencas), 3):
        grupo = sentencas[i:i+3]
        chunks.append(" ".join(grupo))
    return chunks
```

```python
def gerar_chunks_teste_9(texto: str):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=100,
        separators=["\n\n", "\n", " ", ""],
    )
    return splitter.split_text(texto)
```

```python
def gerar_chunks_teste_10(texto_markdown: str):
    headers_to_split_on = [
        ("#", "heading_1"),
        ("##", "heading_2"),
        ("###", "heading_3"),
    ]
    splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
    docs = splitter.split_text(texto_markdown)
    return docs
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

EMBEDDING_MODEL = "sentence-transformers/all-MiniLM-L6-v2"
embedder = HuggingFaceEmbeddings(model_name=EMBEDDING_MODEL)

def gerar_embedding(texto: str):
    return embedder.embed_query(texto)
```

```python
def montar_chunk_dict(chunk_id, document_id, document_name, test_id, strategy,
                       chunk_size, chunk_overlap, texto, embedding, metadata_extra=None):
    metadata = metadata_extra or {}
    return {
        "chunk_id": chunk_id,
        "document_id": document_id,
        "document_name": document_name,
        "test_id": test_id,
        "strategy": strategy,
        "chunk_size": chunk_size,
        "chunk_overlap": chunk_overlap,
        "text": texto,
        "embedding": embedding,
        "metadata": metadata,
    }
```

```python
import json

def processar_documento(nome_doc: str, texto_md: str, pasta_resultados: str):
    resultados_resumo = []   # para o summary.json

    for cfg in TESTES:
        test_id = cfg["test_id"]

        # 1) gerar os textos dos chunks conforme o teste
        if test_id in [1, 2, 3, 4]:
            textos = gerar_chunks_teste_1_a_4(texto_md, cfg["chunk_size"])
            metadatas = [{} for _ in textos]
        elif test_id in [5, 6]:
            textos = gerar_chunks_teste_5_e_6(texto_md, cfg["chunk_overlap"])
            metadatas = [{} for _ in textos]
        elif test_id == 7:
            textos = gerar_chunks_teste_7(texto_md)
            metadatas = [{} for _ in textos]
        elif test_id == 8:
            textos = gerar_chunks_teste_8(texto_md)
            metadatas = [{} for _ in textos]
        elif test_id == 9:
            textos = gerar_chunks_teste_9(texto_md)
            metadatas = [{} for _ in textos]
        elif test_id == 10:
            docs = gerar_chunks_teste_10(texto_md)
            textos = [d.page_content for d in docs]
            metadatas = [d.metadata for d in docs]

        # 2) gerar embedding + montar o JSON de cada chunk
        chunks_json = []
        tamanhos = []
        for i, (texto_chunk, meta) in enumerate(zip(textos, metadatas), start=1):
            if not texto_chunk.strip():
                continue
            emb = gerar_embedding(texto_chunk)
            chunk_id = f"{nome_doc}_test{test_id:02d}_chunk{i:03d}"
            chunk_dict = montar_chunk_dict(
                chunk_id=chunk_id,
                document_id=nome_doc,
                document_name=f"{nome_doc}.pdf",
                test_id=test_id,
                strategy=cfg["strategy"],
                chunk_size=cfg["chunk_size"],
                chunk_overlap=cfg["chunk_overlap"],
                texto=texto_chunk,
                embedding=emb,
                metadata_extra=meta,
            )
            chunks_json.append(chunk_dict)
            tamanhos.append(len(texto_chunk))

        # 3) salvar o arquivo chunks_embeddings.json do teste
        pasta_teste = os.path.join(pasta_resultados, nome_doc, f"test_{test_id:02d}")
        os.makedirs(pasta_teste, exist_ok=True)
        with open(os.path.join(pasta_teste, "chunks_embeddings.json"), "w", encoding="utf-8") as f:
            json.dump(chunks_json, f, ensure_ascii=False, indent=2)

        # 4) calcular estatísticas para o summary
        num_chunks = len(chunks_json)
        avg_size = sum(tamanhos) / num_chunks if num_chunks else 0
        resultados_resumo.append({
            "test_id": test_id,
            "strategy": cfg["strategy"],
            "chunk_size": cfg["chunk_size"],
            "chunk_overlap": cfg["chunk_overlap"],
            "num_chunks": num_chunks,
            "avg_chunk_size": round(avg_size, 1),
            "min_chunk_size": min(tamanhos) if tamanhos else 0,
            "max_chunk_size": max(tamanhos) if tamanhos else 0,
            "embedding_dimension": len(chunks_json[0]["embedding"]) if chunks_json else 0,
        })

        print(f"  [{nome_doc}] teste {test_id:02d} ({cfg['strategy']}): {num_chunks} chunks")

    return resultados_resumo
```

```python
import tiktoken
import json

encoding = tiktoken.get_encoding("cl100k_base")

def contar_tokens(texto):
    # disallowed_special=() faz o tiktoken tratar qualquer coisa como texto normal,
    # mesmo que pareça um token especial como <|endofprompt|>
    return len(encoding.encode(texto, disallowed_special=()))

linhas_tokens = []

for doc in summary:
    nome_doc = doc["document"].replace(".pdf", "")
    for exp in doc["experiments"]:
        if exp["test_id"] in [1, 2, 3, 4, 5, 6]:
            caminho = f"{PASTA_RESULTADOS}/{nome_doc}/test_{exp['test_id']:02d}/chunks_embeddings.json"
            with open(caminho, encoding="utf-8") as f:
                chunks = json.load(f)

            tokens_por_chunk = [contar_tokens(c["text"]) for c in chunks]
            media_tokens = sum(tokens_por_chunk) / len(tokens_por_chunk) if tokens_por_chunk else 0

            linhas_tokens.append({
                "documento": nome_doc,
                "teste": exp["test_id"],
                "chunk_size": exp["chunk_size"],
                "num_chunks": exp["num_chunks"],
                "media_tokens": round(media_tokens, 1),
            })

    print(f"{nome_doc} processado.")

print("\nConcluído! Veja df_tokens abaixo.")
```
