```python
# ============================================================
# 10 ESTRATÉGIAS DE CHUNKING
# ============================================================

# ------------------------------------------------------------
# TESTES 1 A 6 — CHUNKING FIXO
# ------------------------------------------------------------

def chunk_fixo(texto, tamanho, overlap):
    splitter = CharacterTextSplitter(
        separator="",
        chunk_size=tamanho,
        chunk_overlap=overlap,
        length_function=len,
        keep_separator=False
    )

    return splitter.split_text(texto)


def grupo_1(texto):
    return chunk_fixo(texto, 200, 0)


def grupo_2(texto):
    return chunk_fixo(texto, 500, 0)


def grupo_3(texto):
    return chunk_fixo(texto, 1000, 0)


def grupo_4(texto):
    return chunk_fixo(texto, 2000, 0)


def grupo_5(texto):
    return chunk_fixo(texto, 500, 50)


def grupo_6(texto):
    return chunk_fixo(texto, 500, 200)


# ------------------------------------------------------------
# TESTE 7 — POR PARÁGRAFO
# ------------------------------------------------------------

def grupo_7(texto):
    texto = texto.replace("\r\n", "\n")

    paragrafos = re.split(r"\n\s*\n+", texto.strip())

    return [
        paragrafo.strip()
        for paragrafo in paragrafos
        if paragrafo.strip()
    ]


# ------------------------------------------------------------
# TESTE 8 — POR SENTENÇA, AGRUPANDO 3
# ------------------------------------------------------------

def grupo_8(texto):
    texto = texto.replace("\r\n", "\n").strip()

    sentencas = re.split(
        r"(?<=[.!?])\s+",
        texto
    )

    sentencas = [
        sentenca.strip()
        for sentenca in sentencas
        if sentenca.strip()
    ]

    chunks = []

    for i in range(0, len(sentencas), 3):
        grupo = sentencas[i:i + 3]

        chunks.append(
            " ".join(grupo)
        )

    return chunks


# ------------------------------------------------------------
# TESTE 9 — RECURSIVO / SEPARADORES HIERÁRQUICOS
# ------------------------------------------------------------

def grupo_9(texto):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=0,
        length_function=len,
        separators=[
            "\n\n",
            "\n",
            " ",
            ""
        ]
    )

    return splitter.split_text(texto)


# ------------------------------------------------------------
# TESTE 10 — SEÇÕES / HEADINGS MARKDOWN
# ------------------------------------------------------------

def grupo_10(texto):

    # Alguns headings dos arquivos estão escritos como:
    # \## Título
    #
    # O splitter Markdown espera:
    # ## Título
    #
    # Então corrigimos SOMENTE a marcação do heading
    # para que o splitter consiga reconhecê-la.

    texto_normalizado = re.sub(
        r"^(\\)(#{1,6})(\s+)",
        r"\2\3",
        texto,
        flags=re.MULTILINE
    )

    headers_to_split_on = [
        ("#", "Título"),
        ("##", "Seção"),
        ("###", "Subseção"),
        ("####", "Subsubseção")
    ]

    splitter = MarkdownHeaderTextSplitter(
        headers_to_split_on=headers_to_split_on,
        strip_headers=False
    )

    documentos = splitter.split_text(texto_normalizado)

    return [
        documento.page_content.strip()
        for documento in documentos
        if documento.page_content.strip()
    ]


# ------------------------------------------------------------
# ORGANIZAÇÃO DOS 10 GRUPOS
# ------------------------------------------------------------

estrategias = {
    1: grupo_1,
    2: grupo_2,
    3: grupo_3,
    4: grupo_4,
    5: grupo_5,
    6: grupo_6,
    7: grupo_7,
    8: grupo_8,
    9: grupo_9,
    10: grupo_10
}


nomes_estrategias = {
    1: "Fixo - 200 caracteres - overlap 0",
    2: "Fixo - 500 caracteres - overlap 0",
    3: "Fixo - 1000 caracteres - overlap 0",
    4: "Fixo - 2000 caracteres - overlap 0",
    5: "Fixo - 500 caracteres - overlap 50",
    6: "Fixo - 500 caracteres - overlap 200",
    7: "Por parágrafo",
    8: "Por sentença - agrupando 3",
    9: "Recursivo - separadores hierárquicos",
    10: "Por seção - Markdown Headers"
}

print("✅ As 10 estratégias foram definidas!")
```

```python
def buscar_top_3(
    pergunta_embedding,
    chunks,
    embeddings_chunks
):

    resultados = []

    for chunk_id, embedding_chunk in enumerate(
        embeddings_chunks
    ):

        similaridade = similaridade_cosseno(
            pergunta_embedding,
            embedding_chunk
        )

        resultados.append({
            "chunk_id": chunk_id,
            "similaridade": similaridade,
            "chunk": chunks[chunk_id]
        })

    resultados.sort(
        key=lambda x: x["similaridade"],
        reverse=True
    )

    return resultados[:3]
```
