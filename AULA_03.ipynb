## 1. Distância Euclidiana e Distância de Cosseno

Funções pedidas no enunciado, definidas uma única vez e reaproveitadas no resto do notebook.

```python
# Funções de distância e similaridade

import numpy as np

def distancia_euclidiana(vec1: np.ndarray, vec2: np.ndarray) -> float:
    """
    Calcula a Distância Euclidiana entre dois vetores.
    """
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)

    return float(np.linalg.norm(vec1 - vec2))


def similaridade_cosseno(vec1: np.ndarray, vec2: np.ndarray) -> float:
    """
    Calcula a Similaridade de Cosseno entre dois vetores.
    """
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)

    norm_v1 = np.linalg.norm(vec1)
    norm_v2 = np.linalg.norm(vec2)

    if norm_v1 == 0 or norm_v2 == 0:
        return 0.0

    return float(
        np.dot(vec1, vec2) / (norm_v1 * norm_v2)
    )


def distancia_cosseno(vec1: np.ndarray, vec2: np.ndarray) -> float:
    """
    Calcula a Distância de Cosseno.
    """
    return float(1.0 - similaridade_cosseno(vec1, vec2))


print("✅ Funções criadas com sucesso!")
```

## 2. Testando as funções

Usando os vetores de exemplo do enunciado: `embedding_a=[1,0,0]`, `embedding_b=[0,1,0]`, `embedding_c=[1,0,0]`.

```python
# Embeddings de teste fornecidos pelo professor

embedding_a = np.array([1, 0, 0], dtype=float)
embedding_b = np.array([0, 1, 0], dtype=float)
embedding_c = np.array([1, 0, 0], dtype=float)

# Distâncias solicitadas

print("A x B")
print("Distância Euclidiana:", distancia_euclidiana(embedding_a, embedding_b))
print("Similaridade de Cosseno:", similaridade_cosseno(embedding_a, embedding_b))
print("Distância de Cosseno:", distancia_cosseno(embedding_a, embedding_b))

print("\nA x C")
print("Distância Euclidiana:", distancia_euclidiana(embedding_a, embedding_c))
print("Similaridade de Cosseno:", similaridade_cosseno(embedding_a, embedding_c))
print("Distância de Cosseno:", distancia_cosseno(embedding_a, embedding_c))

print("\nB x C")
print("Distância Euclidiana:", distancia_euclidiana(embedding_b, embedding_c))
print("Similaridade de Cosseno:", similaridade_cosseno(embedding_b, embedding_c))
print("Distância de Cosseno:", distancia_cosseno(embedding_b, embedding_c))
```

### 3.2 Visualização MDS (2D e 3D)

Reduz os embeddings de 384 dimensões para 2D/3D para visualizar a proximidade semântica entre os termos.

```python
# MDS 2D - Visualização da proximidade entre termos

import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import MDS

# Termos da atividade
palavras = [
    "gato",
    "felino",
    "cachorro",
    "carro",
    "caminhao",
    "moto",
    "banana",
    "maça",
    "goiaba"
]

# Gerar embeddings
vetores = np.array([
    model.encode(p)
    for p in palavras
])

print("Quantidade de termos:", len(palavras))
print("Dimensão dos embeddings:", vetores.shape[1])

# Reduzir de 384 dimensões para 2
mds = MDS(
    n_components=2,
    dissimilarity="euclidean",
    random_state=42
)

coords = mds.fit_transform(vetores)

# Criar gráfico
plt.figure(figsize=(9, 7))

for i, palavra in enumerate(palavras):
    x, y = coords[i]

    plt.scatter(x, y, s=120)
    plt.annotate(
        palavra,
        (x, y),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center",
        fontsize=11
    )

plt.title("Proximidade de Termos Baseada na Similaridade (MDS)")
plt.xlabel("Componente 1 (MDS)")
plt.ylabel("Componente 2 (MDS)")
plt.tight_layout()
plt.show()
```

```python
# MDS 3D - Visualização da proximidade entre termos

import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import MDS

# Termos da atividade
palavras = [
    "gato",
    "felino",
    "cachorro",
    "carro",
    "caminhao",
    "moto",
    "banana",
    "maça",
    "goiaba"
]

# Gerar embeddings
vetores = np.array([
    model.encode(p)
    for p in palavras
])

print("Quantidade de termos:", len(palavras))
print("Dimensão dos embeddings:", vetores.shape[1])

# Reduzir de 384 dimensões para 3
mds = MDS(
    n_components=3,
    dissimilarity="euclidean",
    random_state=42
)

coords_3d = mds.fit_transform(vetores)

# Criar gráfico 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

for i, palavra in enumerate(palavras):
    x, y, z = coords_3d[i]

    ax.scatter(x, y, z, s=120)
    ax.text(x, y, z + 0.1, palavra, fontsize=10)

ax.set_title(
    "Proximidade de Termos Baseada na Similaridade em 3D (MDS)"
)

ax.set_xlabel("Componente 1 (MDS)")
ax.set_ylabel("Componente 2 (MDS)")
ax.set_zlabel("Componente 3 (MDS)")

plt.tight_layout()
plt.show()
```

## 4. Busca Semântica Simples Manual (Embedding por Embedding)

A partir daqui, seguimos os passos do enunciado:
1. Ler os arquivos `.md`
2. Separar o texto (linha, depois parágrafo, depois capítulo)
3. Gerar embedding de cada trecho
4. Definir perguntas de busca
5. Calcular similaridade de cosseno da query contra cada trecho
6. Retornar o TOP 3

### 4.1 Lendo os arquivos markdown

```python
import os

arquivos_md = [
    arquivo
    for arquivo in os.listdir()
    if arquivo.endswith(".md")
]

print("Arquivos .md encontrados:")

for arquivo in arquivos_md:
    print("✅", arquivo)
```

```python
import os

conteudos = {}

for arquivo in arquivos_md:
    with open(arquivo, "r", encoding="utf-8") as f:
        conteudos[arquivo] = f.read()

print("📚 Documentos carregados:")
print()

for nome, texto in conteudos.items():
    print(f"✅ {nome}")
    print(f"   Quantidade de caracteres: {len(texto)}")
```

```python
for nome, texto in conteudos.items():
    print("=" * 80)
    print(f"📄 {nome}")
    print("=" * 80)
    print(texto[:1000])
    print("\n")
```

### 4.2 Dividindo por linha e gerando embeddings

```python
# Dividindo os documentos linha por linha

linhas_documentos = []

for nome_arquivo, texto in conteudos.items():
    linhas = texto.splitlines()

    for numero_linha, linha in enumerate(linhas, start=1):
        linha = linha.strip()

        # Ignorar linhas vazias
        if linha:
            linhas_documentos.append({
                "documento": nome_arquivo,
                "linha": numero_linha,
                "texto": linha
            })

print(f"✅ Total de linhas/trechos encontrados: {len(linhas_documentos)}")
```

```python
import pandas as pd

df_linhas = pd.DataFrame(linhas_documentos)

df_linhas.head(20)
```

```python
# Remover linhas que representam imagens ou conteúdo vazio

df_linhas = df_linhas[
    ~df_linhas["texto"].str.strip().str.lower().eq("<!-- image -->")
].copy()

df_linhas = df_linhas.reset_index(drop=True)

print(f"✅ Total de trechos após a limpeza: {len(df_linhas)}")

df_linhas.head(20)
```

```python
# Gerando embeddings para todas as linhas dos documentos

embeddings_linhas = model.encode(
    df_linhas["texto"].tolist(),
    show_progress_bar=True
)

print("✅ Embeddings gerados com sucesso!")
print("Quantidade de embeddings:", len(embeddings_linhas))
print("Dimensão de cada embedding:", embeddings_linhas.shape[1])
```

### 4.3 Busca semântica por linha (exemplo com 1 query)

```python
# Pergunta para a busca semântica

query = "Quais são os impactos da inteligência artificial na sociedade?"

# Gerar o embedding da pergunta
embedding_query = model.encode(query)

# Calcular a similaridade da pergunta com todos os trechos
scores = []

for embedding in embeddings_linhas:
    score = similaridade_cosseno(embedding_query, embedding)
    scores.append(score)

# Adicionar os scores ao DataFrame
df_linhas["similaridade"] = scores

# Ordenar do mais semelhante para o menos semelhante
df_resultado = df_linhas.sort_values(
    by="similaridade",
    ascending=False
).reset_index(drop=True)

# Mostrar os 3 trechos mais semelhantes
df_resultado.head(3)
```

```python
# Mostrar os resultados da busca semântica de forma organizada

for i, row in df_resultado.head(3).iterrows():
    print("=" * 80)
    print(f"RESULTADO {i + 1}")
    print(f"Documento: {row['documento']}")
    print(f"Linha: {row['linha']}")
    print(f"Similaridade: {row['similaridade']:.4f}")
    print(f"Texto: {row['texto']}")
```

### 4.4 Dividindo por parágrafo e gerando embeddings

(Passo 7 do enunciado)

```python
# Criando parágrafos a partir dos arquivos originais

paragrafos = []

for documento, texto in conteudos.items():

    # Divide usando duas quebras de linha
    partes = texto.split("\n\n")

    numero_paragrafo = 0

    for parte in partes:
        parte = parte.strip()

        # Ignora trechos vazios ou muito pequenos
        if len(parte) < 25:
            continue

        paragrafos.append({
            "documento": documento,
            "paragrafo": numero_paragrafo,
            "texto": parte
        })

        numero_paragrafo += 1

df_paragrafos = pd.DataFrame(paragrafos)

print("✅ Parágrafos criados!")
print("Quantidade de parágrafos:", len(df_paragrafos))

display(df_paragrafos.head(10))
```

```python
# Gerando embeddings para todos os parágrafos

embeddings_paragrafos = model.encode(
    df_paragrafos["texto"].tolist(),
    show_progress_bar=True
)

print("✅ Embeddings dos parágrafos gerados!")
print("Quantidade de embeddings:", len(embeddings_paragrafos))
print("Dimensão de cada embedding:", embeddings_paragrafos.shape[1])
```

```python
# Busca semântica por parágrafos

query = "Quais são os impactos da inteligência artificial na sociedade?"

# Gerar embedding da pergunta
embedding_query = model.encode(query)

# Calcular similaridade com todos os parágrafos
scores_paragrafos = []

for embedding in embeddings_paragrafos:
    score = similaridade_cosseno(embedding_query, embedding)
    scores_paragrafos.append(score)

# Adicionar os scores ao DataFrame
df_paragrafos["similaridade"] = scores_paragrafos

# Ordenar pelos maiores scores
df_resultado_paragrafos = (
    df_paragrafos
    .sort_values(by="similaridade", ascending=False)
    .reset_index(drop=True)
)

# Mostrar TOP 3
df_resultado_paragrafos.head(3)
```

### 4.5 Dividindo por capítulo/seção e gerando embeddings

(Passo 8 do enunciado)

```python
# Criando capítulos/seções a partir dos títulos Markdown (##)

capitulos = []

for documento, texto in conteudos.items():

    # Divide o documento sempre que encontrar "## "
    partes = texto.split("## ")

    numero_capitulo = 0

    for parte in partes:
        parte = parte.strip()

        # Ignora trechos muito pequenos
        if len(parte) < 25:
            continue

        capitulos.append({
            "documento": documento,
            "capitulo": numero_capitulo,
            "texto": parte
        })

        numero_capitulo += 1

df_capitulos = pd.DataFrame(capitulos)

print("✅ Capítulos/seções criados!")
print("Quantidade de capítulos/seções:", len(df_capitulos))

display(df_capitulos.head(10))
```

```python
# Gerando embeddings para todos os capítulos/seções

embeddings_capitulos = model.encode(
    df_capitulos["texto"].tolist(),
    show_progress_bar=True
)

print("✅ Embeddings dos capítulos/seções gerados!")
print("Quantidade de embeddings:", len(embeddings_capitulos))
print("Dimensão de cada embedding:", embeddings_capitulos.shape[1])
```

```python
# Busca semântica por capítulos/seções

query = "Quais são os impactos da inteligência artificial na sociedade?"

# Gerar embedding da pergunta
embedding_query = model.encode(query)

# Calcular similaridade com todos os capítulos/seções
scores_capitulos = []

for embedding in embeddings_capitulos:
    score = similaridade_cosseno(embedding_query, embedding)
    scores_capitulos.append(score)

# Adicionar os scores ao DataFrame
df_capitulos["similaridade"] = scores_capitulos

# Ordenar pelos maiores scores
df_resultado_capitulos = (
    df_capitulos
    .sort_values(by="similaridade", ascending=False)
    .reset_index(drop=True)
)

# Mostrar TOP 3
df_resultado_capitulos.head(3)
```

### 4.6 Rodando as perguntas da atividade nas 3 granularidades

('O que é Autonomia e opacidade algorítmica?', 'O que é o diário de bordo da IA?', etc.)

```python
# Perguntas propostas na atividade

perguntas = [
    "O que é Autonomia e opacidade algorítmica?",
    "O que é o diário de bordo da IA?",
    "Qual o impacto dos algoritmos nas redes sociais e no espaço público?"
]

for pergunta in perguntas:
    print("=" * 100)
    print(f"🔍 PERGUNTA: {pergunta}")
    print("=" * 100)

    # Embedding da pergunta
    embedding_query = model.encode(pergunta)

    # -------------------------------
    # 1. BUSCA POR LINHAS
    # -------------------------------
    scores = [
        similaridade_cosseno(embedding_query, emb)
        for emb in embeddings_linhas
    ]

    resultado_linhas = df_linhas.copy()
    resultado_linhas["similaridade"] = scores

    resultado_linhas = (
        resultado_linhas
        .sort_values("similaridade", ascending=False)
        .head(3)
    )

    print("\n📌 1. BUSCA POR LINHAS — TOP 3")
    display(resultado_linhas)

    # -------------------------------
    # 2. BUSCA POR PARÁGRAFOS
    # -------------------------------
    scores = [
        similaridade_cosseno(embedding_query, emb)
        for emb in embeddings_paragrafos
    ]

    resultado_paragrafos = df_paragrafos.copy()
    resultado_paragrafos["similaridade"] = scores

    resultado_paragrafos = (
        resultado_paragrafos
        .sort_values("similaridade", ascending=False)
        .head(3)
    )

    print("\n📌 2. BUSCA POR PARÁGRAFOS — TOP 3")
    display(resultado_paragrafos)

    # -------------------------------
    # 3. BUSCA POR CAPÍTULOS / SEÇÕES
    # -------------------------------
    scores = [
        similaridade_cosseno(embedding_query, emb)
        for emb in embeddings_capitulos
    ]

    resultado_capitulos = df_capitulos.copy()
    resultado_capitulos["similaridade"] = scores

    resultado_capitulos = (
        resultado_capitulos
        .sort_values("similaridade", ascending=False)
        .head(3)
    )

    print("\n📌 3. BUSCA POR CAPÍTULOS / SEÇÕES — TOP 3")
    display(resultado_capitulos)
```
